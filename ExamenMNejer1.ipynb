```python
import numpy as np
from scipy.optimize import newton, bisect

def f(x):
    return x**5 + 5.25*x**4 + 4.125*x**3 - 9.125*x**2 - 14.625*x - 5.625

def fprime(x):
    return 5*x**4 + 21*x**3 + 12.375*x**2 - 18.25*x - 14.625

roots_newton = []
for guess in [-2, -1, -0.5, 0, 1, 2]:
    try:
        r = newton(f, guess, fprime, tol=1e-8, maxiter=50)
        roots_newton.append(round(r, 6))
    except:
        pass

roots_newton = sorted(set(roots_newton))

coef = [1, 5.25, 4.125, -9.125, -14.625, -5.625]
roots_all = np.roots(coef)

print("Raíces reales Newton:", roots_newton)
print(" raíces ")
for r in roots_all:
    print(f"{r:.6f}")
```
